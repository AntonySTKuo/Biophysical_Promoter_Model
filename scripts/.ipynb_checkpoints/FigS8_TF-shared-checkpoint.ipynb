```python
## === Core Libraries ===
import numpy as np
import pandas as pd
from copy import copy

## === Statistics ===
from scipy.stats import pearsonr

## === Visualization ===
import matplotlib.pyplot as plt
from matplotlib.ticker import FormatStrFormatter
import logomaker

## === My Modules ===
from util import load_obj, save_obj, sci_ticks, showText, PWM2Score, Motif2Seqs
```

```python
def load_df(lib_name):
    """Load data from pkl file depending on library name"""
    paths = {
        'TetR': "PL_TetR",
        'LuxR': "PL_LuxR",
        'CueR': "PL_CueR"
        }

    if lib_name not in paths:
        raise ValueError(f"Unknown library: {lib_name}")

    lib = paths[lib_name]
    dfB = pd.read_pickle(f"../tables/{lib}b.pkl")  # Baseline condition
    dfI = pd.read_pickle(f"../tables/{lib}i.pkl")  # Induced condition

    return dfB, dfI
```

```python
def drawHist(ax, df, color='#999999', alpha=1):
    weights = np.ones_like(df['LogGFP']) / len(df['LogGFP']) * 100
    ax.hist(df['LogGFP'], bins=np.linspace(0, 3.2, 11), edgecolor='black', linewidth=0.5, 
            color=color, weights=weights, alpha=alpha)
    return None
```

## ==== Fig 6D: Basal vs. Induced promoter strength ====

```python
def showFC(ax, df, lib):
    """
    Plot the scatter points for function vs. binding for a given transcription factor library.

    Parameters:
        ax (matplotlib.axes.Axes): The axes to plot on.
        df (pandas.DataFrame): DataFrame containing 'B' (basal expression) and 'F' (fold change) data.
        lib (str): Library name, e.g., "TetR", "LuxR", or "CueR".
    """
    
    ## Plot all data points in gray
    if lib == 'TetR...':
        x, y = df['B'], df['I']
    else:
        x, y = df['B'], df['I']
    ax.scatter(x, y, s=0.5, alpha=0.15, color='0.55', edgecolor='None')
    
    ## Highlight the wild type variants
    wild_type = {'TetR': 'TTGacaTaTTTt', 'LuxR': 'tTTacGTaTAGt', 'CueR': 'TTGaccTaACCt'}
    ax.scatter(x.loc[wild_type[lib]], y.loc[wild_type[lib]], s=8, color='red')
    percentile = df['F'].rank(pct=True).loc[wild_type[lib]] * 100
    # print(f"WT% in PL_{lib} = {100-percentile}; FC = {10**df['F'].loc[wild_type[lib]]}")
    
    ## Highlight the consensus variants
    consensus = {'TetR': 'TTGacaTaTAAt', 'LuxR': 'tTGacATaTAAt', 'CueR': 'TTGaccTaTAAt'}
    ax.scatter(x.loc[consensus[lib]], y.loc[consensus[lib]], s=8, color='blue')
    percentile = df['F'].rank(pct=True).loc[consensus[lib]] * 100
    # print(f"CS% in PL_{lib} = {100-percentile}; FC = {10**df['F'].loc[consensus[lib]]}")

    ## Highlight the PLLuxR HF variant
    if lib == 'LuxR':
        vairnatHF = {'LuxR': 'tCCacTTaTATt'}
        ax.scatter(x.loc[vairnatHF[lib]], y.loc[vairnatHF[lib]], s=8, color='#895129')
        percentile = df['F'].rank(pct=True).loc[vairnatHF[lib]] * 100
        # print(f"HF% in PL_{lib} = {100-percentile}; FC = {10**df['F'].loc[vairnatHF[lib]]}")
        # print(f"{df['F'].rank(ascending=False, pct=False).loc[vairnatHF[lib]]}")
        
    ## XY-axis limits
    ax.set_xlim(np.min(x)-.1, np.max(x)+.2)
    ax.set_ylim(np.min(y)-.1, np.max(y)+.2)
    
    return df['B'], df['I'], df['F']
```

```python
libs = ["TetR", "LuxR", "CueR"]
fig, axes = plt.subplots(1, 3, figsize=(80/25.4, 22/25.4), dpi=600, sharey=False)

for ax, lib in zip(axes, libs):
    
    dfB, dfI = load_df(lib)
    df = pd.DataFrame({"I": dfI['LogGFP'], "B": dfB['LogGFP']})
    df['F'] = df['I'] - df['B']
    B, I, F = showFC(ax, df, lib)
    
    sci_ticks(ax, 'x', ticksize=5, pad=1.5)
    sci_ticks(ax, 'y', ticksize=5, pad=1.2)
    ax.tick_params(axis='both', which='both', length=0)


## ==== Final adjustments and saving ====
for ax in axes:
    for spine in ax.spines.values():
        spine.set_linewidth(0.4)

plt.subplots_adjust(left=0.10, right=0.95, bottom=0.15, top=0.95, wspace=0.4)
# plt.savefig("../figures/Fig6D_Basal-Induced.png")
# plt.savefig("../figures/Fig6D_Basal-Induced.svg")
plt.show()
```

## ==== Fig 6E: Energy-to-Strength curve ====

```python
def stratified_sample(df, col, counts, seed=77777):
    rs = np.random.RandomState(seed)
    d = df.copy()
    d["grp"] = pd.cut(d[col], bins=len(counts))
    parts = []
    for grp, cnt in zip(d["grp"].unique(), counts):
        block = d[d["grp"] == grp]
        if block.empty:
            continue
        parts.append(block.sample(cnt, replace=True, random_state=seed))
    out = pd.concat(parts).drop(columns="grp")
    return out

def metrics(y, yhat):
    y, yhat = np.asarray(y, float).ravel(), np.asarray(yhat, float).ravel()
    r = np.corrcoef(y, yhat)[0,1] if y.size>1 else np.nan
    sse = np.sum((y-yhat)**2); sst = np.sum((y-y.mean())**2)
    r2 = np.nan if np.isclose(sst,0) else 1 - sse/sst
    mae = np.mean(np.abs(y-yhat))
    return {"r2": r2, "pearson_r": r, "mae": mae}

def extract_uppercases(input_string):
    uppercase_letters = [char for char in input_string if char.isupper()]
    return ''.join(uppercase_letters)
```

```python
## Sigmoid Curve ====
def E2E(dGs, params, beta=1.623):
    e0 = params['e0']
    rmin = params['rmin']
    rmax = params['rmax']
    
    bound_RNAP = np.exp(-beta*(dGs+e0))
    prob = (bound_RNAP) / (1+bound_RNAP)
    expr = rmax*prob + rmin
    
    return np.log10(expr)

def SigmoidCurve(ax, dGs, expression, params, lib, c_line='#666666', c_point='#b3b3b3'):
    
    if c_line == '#666666':
        xs = {
            'TetR': np.linspace(-11, 9, 101), 
            'LuxR': np.linspace(-11, 12, 101), 
            'CueR': np.linspace(-11, 13, 101)
            }[lib]
    else:
        xs = {
            'TetR': np.linspace(-10, 10, 101), 
            'LuxR': np.linspace( -7, 16, 101), 
            'CueR': np.linspace(-14, 10, 101)
            }[lib]
    ys = E2E(xs, params)
    
    ax.scatter(-dGs, expression, s=1, alpha=0.6, edgecolor='None', color=c_point)
    ax.plot(-xs, ys, color=c_line, linewidth=0.8)
    ax.set_xticks([-(xs[0]+6), -(xs[-1]-6)])
    
    ## outlier (shifted -10)
    if lib == 'TetR':
        m = {'TetR': 'NNNacaAaNNTt', 'LuxR': 'tTGacNGaTANt', 'CueR': 'TTGaccAaNNTt'}
        ax.scatter(-dGs.loc[Motif2Seqs(m[lib])], expression.loc[Motif2Seqs(m[lib])], s=1, edgecolor='None', color='#edce5f')
    
    ## Label specific variants
    wild_type = {'TetR': 'TTGacaTaTTTt', 'LuxR': 'tTTacGTaTAGt', 'CueR': 'TTGaccTaACCt'}
    ax.scatter(-dGs.loc[wild_type[lib]], expression.loc[wild_type[lib]], s=10, color='red' , edgecolor='black', linewidth=0.2, zorder=2)
    
    ## Label specific variants
    consensus = {'TetR': 'TTGacaTaTAAt', 'LuxR': 'tTGacATaTAAt', 'CueR': 'TTGaccTaTAAt'}
    ax.scatter(-dGs.loc[consensus[lib]], expression.loc[consensus[lib]], s=10, color='blue', edgecolor='black', linewidth=0.2, zorder=2)
    
    return None
```

```python
libs = ["TetR", "LuxR", "CueR"]
fig, axes = plt.subplots(3, 2, figsize=(47/25.4, 77/25.4), dpi=300, sharey='row')

for axs, lib in zip(axes, libs):

    dfB, dfI = load_df(lib)
    df = pd.DataFrame({"I": dfI['LogGFP'], "B": dfB['LogGFP']})

    if lib == 'CueR':
        motifs = {"TetR": "NNNacaAaNNTt", "LuxR": "tTGacNGaTANt", "CueR": "TTGaccAaNNTt"}
        df.drop(index=list(Motif2Seqs(motifs[lib])), inplace=True)
    
    paramsB = load_obj(f"../models/Params_{lib}b.pkl")
    paramsI = load_obj(f"../models/Params_{lib}i.pkl")
    dG_listB = PWM2Score(paramsB['pwm'])
    dG_listI = PWM2Score(paramsI['pwm'])
    df['dG_B'] = np.array([dG_listB[s] for s in df.index.map(extract_uppercases)])
    df['dG_I'] = np.array([dG_listI[s] for s in df.index.map(extract_uppercases)])
    
    axB, axI = axs
    maeB = SigmoidCurve(axB, df['dG_B'], df['B'], paramsB, lib, c_line='#666666', c_point='#b3b3b3')
    maeI = SigmoidCurve(axI, df['dG_I'], df['I'], paramsI, lib, c_line='#2aa22a', c_point='#85e085')
    
    axB.xaxis.set_major_formatter(FormatStrFormatter('%.0f'))
    axI.xaxis.set_major_formatter(FormatStrFormatter('%.0f'))
    axB.tick_params(axis='x', which='both', direction='in', labelsize=5, width=0.3, length=2, pad=1.2)
    axI.tick_params(axis='x', which='both', direction='in', labelsize=5, width=0.3, length=2, pad=1.2)
    
    dfB = stratified_sample(df, 'B', [1000]*8)
    dfI = stratified_sample(df, 'I', [1000]*8)

for ax in axes.flatten():
    sci_ticks(ax, 'y', ticksize=5, pad=1.5)
    for spine in ax.spines.values():
        spine.set_linewidth(0.4)

    yMin, yMax = ax.set_ylim()
    ax.set_ylim(yMin, yMax + (yMax-yMin)*0.06)

plt.subplots_adjust(left=0.20, right=0.95, bottom=0.20, top=0.95, wspace=0, hspace=0.4)
# plt.savefig("../figures/Fig6_RegFitting_E2Ecurve.svg")
# plt.savefig("../figures/Fig6_RegFitting_E2Ecurve.png")
plt.show()
```

## ==== Fig 6F: Energy logo ====

```python
def pwm4logo(df, positions, nucleotides):

    def insert_row(df, new_row, position):
        if isinstance(new_row, dict):
            new_row = pd.DataFrame(new_row, index=[position])
        elif isinstance(new_row, list):
            new_row = pd.DataFrame([new_row], columns=df.columns, index=[position])
        else:
            raise ValueError("Invalid input format. Expected dictionary or list.")

        df_before = df.iloc[:position]
        df_after = df.iloc[position:]
        df = pd.concat([df_before, new_row, df_after]).reset_index(drop=True)
        return df

    df[u'А'] = 0
    df[u'С'] = 0
    df[u'Т'] = 0

    pseudoN = {
        u'А': [0, 0, 0, 0, -2, 0, 0],
        u'С': [0, 0, 0, 0, 0, -2, 0],
        u'Т': [0, 0, 0, 0, 0, 0, -2],
        }

    for position, N in zip(positions, nucleotides):
        df = insert_row(df, pseudoN[N], position)

    return df

def pwm4logo_TF(pwm, mode='TetR'):
    return {
        'TetR': pwm4logo(pwm, [3, 4, 5, 7, 11], u'АСААТ'),
        'LuxR': pwm4logo(pwm, [0, 3, 4, 7, 11], u'ТАСАТ'),
        'CueR': pwm4logo(pwm, [3, 4, 5, 7, 11], u'АССАТ')}[mode]

def showLogo(ax, df):
    
    ## custom color scheme
    color_scheme = {
        'A' : 'green',
        'C' : 'blue',
        'G' : 'orange',
        'T' : 'red',
        u'АСᏀТ': '#dfdfdf'
        }

    ## create Logo object
    pwm_logo = logomaker.Logo(df,
                              shade_below=.4,
                              fade_below=.4,
                              font_name='DejaVu Sans',
                              color_scheme=color_scheme,
                              ax=ax)

    ## style using Logo methods
    pwm_logo.style_spines(visible=False)
    pwm_logo.style_spines(spines=['left', 'bottom'], visible=True)

    ax.axhline(0.0, color='black', lw=0.4)
    ax.axvline(5.5, color='black', lw=0.4)
    ax.patch.set_edgecolor('black')
    ax.patch.set_linewidth(0.4)
    
    return pwm_logo
```

```python
libs = ["TetR", "LuxR", "CueR"]
fig, axes = plt.subplots(nrows=2, ncols=3, dpi=300, figsize=(130/25.4, 45/25.4), sharex=True, sharey=True,)

for axs, lib in zip(axes.T, libs):
    paramsB = load_obj(f"../models/Params_{lib}b.pkl")
    paramsI = load_obj(f"../models/Params_{lib}i.pkl")
    pwmB = pwm4logo_TF(copy(paramsB["pwm"]), mode=lib)
    pwmI = pwm4logo_TF(copy(paramsI["pwm"]), mode=lib)
    showLogo(axs[0], pd.concat([-pwmB[ :6], -pwmI[ :6]]).reset_index(drop=True))
    showLogo(axs[1], pd.concat([-pwmB[-6:], -pwmI[-6:]]).reset_index(drop=True))

    for ax in axs:
        ax.set(
            ylim=(-3, 3),
            yticks=[-2, 0, 2],
            xticks=[]
            )
        ax.yaxis.set_major_formatter(FormatStrFormatter('%.0f'))
        ax.tick_params(axis='y', which='both', direction='in', labelsize=5, width=0.3, length=2.1, pad=1.2)
        for spine in ax.spines.values():
            spine.set_linewidth(0.4)


plt.subplots_adjust(left=0.20, right=0.95, bottom=0.2, top=0.95, wspace=0.4, hspace=0.2)
# plt.savefig("../figures/Fig6_Logo.svg")
plt.show()
```

## ==== Fig 6G: Energy comparison ====

```python
def pearsonr_without_nan(x, y):
    valid_indices = ~np.isnan(x) & ~np.isnan(y)
    x_valid = x[valid_indices]
    y_valid = y[valid_indices]
    return pearsonr(x_valid, y_valid)

def calculate_slope_ignore_nan(x, y):
    """
    Calculate the slope of the linear regression line while ignoring NaN values.

    Parameters:
        x (numpy.ndarray): The independent variable (predictor) data.
        y (numpy.ndarray): The dependent variable (target) data.

    Returns:
        float: The slope of the linear regression line while ignoring NaN values.
    """
    mask = ~np.isnan(x) & ~np.isnan(y)
    slope, _ = np.polyfit(x[mask], y[mask], 1)
    return slope

def pwm4comparison(pwm, mode):
    
    if mode == 'TetR':
        pwm35 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm35.loc[[0, 1, 2]] = pwm.loc[[0, 1, 2]].values
        pwm10 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm10.loc[[0, 2, 3, 4]] = pwm.loc[[3, 4, 5, 6]].values
        
    elif mode == 'LuxR':
        pwm35 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm35.loc[[1, 2, 5]] = pwm.loc[[0, 1, 2]].values
        pwm10 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm10.loc[[0, 2, 3, 4]] = pwm.loc[[3, 4, 5, 6]].values
        
    elif mode == 'CueR':
        pwm35 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm35.loc[[0, 1, 2]] = pwm.loc[[0, 1, 2]].values
        pwm10 = pd.DataFrame(np.repeat(np.nan, 24).reshape(6, 4), columns=['A', 'C', 'G', 'T'])
        pwm10.loc[[0, 2, 3, 4]] = pwm.loc[[3, 4, 5, 6]].values
        
    return pwm35, pwm10

def PairwiseComparison(ax, pwm35, pwmC35, pwm10, pwmC10):
    
    def plot_pair(ax, x, y, color, text_y):

        ## Scatter
        ax.scatter(-x, -y, color=color, s=6.5, alpha=0.6, edgecolors='None')

        ## Set limits
        xlim = ax.set_xlim([-3, 2])
        ylim = ax.set_ylim([-2, 3])

        ## Pearson r and display
        r, p = pearsonr_without_nan(x.values, y.values)
        # ax.text(2.8, text_y, fr"$r={r:.3f}, P={p:.3f}$", color=color, fontsize=5, ha='right', va='center')
        # print(f"r={r}\nP={p}\n")

        return ax

    ## Plot -35 pair on main axis
    plot_pair(ax, pwmC35, pwm35, color='blue', text_y=-1.5)
    plot_pair(ax, pwmC10, pwm10, color='red',  text_y=-2.4)

    ## Style
    for spine in ax.spines.values():
        spine.set_linewidth(0.4)

    return None
```

```python
## Define the libraries
libs = ["TetRb", "TetRi", "LuxRb", "LuxRi", "CueRb", "CueRi"]

## Figure layout
fig, axes = plt.subplots(3, 2, figsize=(47/25.4, 90/25.4), dpi=300, sharey=True)

## Load the parameters of the constitutive library
paramsC = load_obj(f"../models/Params_C17.pkl")

## Loop over each library and its corresponding axis
for ax, lib in zip(axes.flatten(), libs):
    params = load_obj(f"../models/Params_{lib}.pkl")
    pwm35, pwm10 = pwm4comparison(copy(params['pwm']), mode=lib[:-1])
    PairwiseComparison(ax, pwm35, paramsC['pwm35'], pwm10, paramsC['pwm10'])
    
    ax.set_xticks([-2, -1,  0,  1])
    ax.set_yticks([-1,  0,  1,  2])
    ax.xaxis.set_major_formatter(FormatStrFormatter('%.0f'))
    ax.yaxis.set_major_formatter(FormatStrFormatter('%.0f'))
    ax.tick_params(axis='both', which='both', direction='in', labelsize=5, width=0.3, length=2.1, pad=1.2)

## Output
plt.subplots_adjust(left=0.20, right=0.95, bottom=0.20, top=0.95, wspace=0, hspace=0.4)
# plt.savefig("../figures/Fig6_RegFitting_Comparison.svg")
plt.show()
```
